```python
#!/usr/bin/env sage
"""
SageMath code to compute all integers c with multiplicity >=3
for the equation F_n - u = c, where:
  - (F_n) is the Fibonacci sequence (F_0=0, F_1=1)
  - u in U_{2} = {+/- 2^e : e >= 0 integer}
"""

def compute_triple_representations(N_max=150, E_max=150):
    """
    Compute all integers c with at least three distinct representations
    F_n - (sign * 2^e) = c for 0 <= n <= N_max, 0 <= e <= E_max.
    
    Returns:
        dict: mapping c -> list of representations [(n, sign, e), ...]
              where sign = +1 for u=+2^e, sign = -1 for u=-2^e
    """
    from sage.all import fibonacci
    
    # Dictionary to accumulate representations: c -> [(n, sign, e), ...]
    rep_dict = {}
    
    # Precompute all Fibonacci numbers up to N_max
    fibs = [fibonacci(n) for n in range(N_max + 1)]
    
    # Iterate over all indices and exponents
    for n in range(N_max + 1):
        F_n = fibs[n]
        for e in range(E_max + 1):
            pow2 = Integer(1) << e  # Fast 2^e computation
            
            # Case 1: u = +2^e  ->  c = F_n - 2^e
            c1 = F_n - pow2
            rep_dict.setdefault(c1, []).append((n, 1, e))
            
            # Case 2: u = -2^e  ->  c = F_n + 2^e
            c2 = F_n + pow2
            rep_dict.setdefault(c2, []).append((n, -1, e))
    
    # Filter to keep only c with >=3 distinct representations
    triples = {c: reps for c, reps in rep_dict.items() if len(reps) >= 3}
    return triples

def format_representation(rep):
    """Format a single representation (n, sign, e) as human-readable string."""
    n, sign, e = rep
    sign_str = "" if sign == 1 else "-"
    return "(n=%d, u=%s2^%d)" % (n, sign_str, e)

def main():
    # Compute all c with multiplicity >=3 (bounds already sufficient for |c|<=200)
    triples = compute_triple_representations(N_max=150, E_max=150)
    
    # Filter to |c| <= 200
    triples_small = {c: reps for c, reps in triples.items() if abs(c) <= 200}
    sorted_triples = sorted(triples_small.items(), key=lambda x: x[0])
    
    print("="*70)
    print("Integers c with >=3 representations F_n - u = c and |c| <= 200")
    print("where u in U_{2} = {+/- 2^e : e >= 0}")
    print("Computed for 0 <= n <= 150, 0 <= e <= 150")
    print("="*70)
    print("%6s | Representations (n, u)" % "c")
    print("-"*70)
    
    for c, reps in sorted_triples:
        rep_strs = [format_representation(r) for r in reps[:5]]
        if len(reps) > 5:
            rep_strs.append("... (+%d more)" % (len(reps)-5))
        print("%6d | " % c + ", ".join(rep_strs))
    
    print("-"*70)
    print("Total integers c with multiplicity >=3 and |c| <= 200: %d" % len(sorted_triples))
    print("="*70)
    
    # Verification
    large_found = [c for c in triples if 100 < abs(c) <= 200]
    if not large_found:
        print("\nVerification: NO integers c with 100 < |c| <= 200 have multiplicity >=3")
        print("All triple-representation c satisfy |c| <= 85")
    else:
        print("\nWarning: Found %d values with 100 < |c| <= 200:" % len(large_found))
        print(large_found)

if __name__ == "__main__":
    main()
```
